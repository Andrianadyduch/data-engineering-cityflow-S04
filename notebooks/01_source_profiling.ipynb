## Дослідження таблиці orders

```python
orders.describe()
```

```python
orders.nunique()
```

```python
print("Order status:")
print(orders["order_status"].value_counts(dropna=False))

print("\nCurrency:")
print(orders["currency"].value_counts(dropna=False))
```

```python
missing_count = orders.isna().sum()
missing_share = orders.isna().mean() * 100

pd.DataFrame({
    "missing_count": missing_count,
    "missing_share_%": missing_share
})
```

```python
text_columns = orders.select_dtypes(include="object").columns

for col in text_columns:
    empty_count = (orders[col] == "").sum()
    whitespace_count = orders[col].astype(str).str.fullmatch(r"\s+").sum()
    
    print(
        col,
        "- empty:", empty_count,
        "- whitespace:", whitespace_count
    )
```

```python
print("Full duplicates:", orders.duplicated().sum())
print("Duplicate order_id:", orders.duplicated(subset=["order_id"]).sum())
```

```python
display(orders[orders.duplicated(keep=False)])

display(
    orders[orders.duplicated(subset=["order_id"], keep=False)]
    .sort_values("order_id")
)
```

```python
orders_dates = pd.to_datetime(orders["created_at"], errors="coerce")

print("Min date:", orders_dates.min())
print("Max date:", orders_dates.max())
print("Invalid dates:", orders_dates.isna().sum())
```

```python
invalid_orders_dates = orders[orders_dates.isna()]

display(invalid_orders_dates[
    ["order_id", "created_at"]
])
```

```python
numeric_columns = [
    "subtotal_amount",
    "delivery_fee",
    "discount_amount",
    "total_amount"
]

for col in numeric_columns:
    print(f"\n{col}")
    print("Min:", orders[col].min())
    print("Max:", orders[col].max())
    print("Zero:", (orders[col] == 0).sum())
    print("Negative:", (orders[col] < 0).sum())
```
